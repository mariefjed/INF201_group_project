## Project Overview
This notebook presents our mini-RPG battle simulator implemented using object-oriented programming in Python. The project covers data encapsulation, custom classes (`Dice`, `Weapon`, `Battle`), inheritance (`Warrior`, `Mage`), and object composition.

```python
from battle import Battle
from character import Character, Mage, Warrior
from dice import Dice
from items import Weapon
```

# Summary of the Game Simulator Project

## 1. Class Overview
* **`Character`**: The base class managing encapsulated health (`_hp`), damage taking, healing, attacking, and properties (`hp`, `is_alive`).
* **`Dice`**: Simulates an n-sided dice using `random.randint`, supporting direct rolling via `roll()` and instance calls (`__call__`).
* **`Weapon`**: Implements object composition by binding a name to a `Dice` object to generate dynamic weapon damage (`roll_damage()`).
* **`Warrior`**: A subclass that inherits from `Character` and overrides `take_damage()` to reduce incoming damage by `armour`.
* **`Mage`**: A subclass that manages a `mana` resource, overriding `attack()` to cast high-damage spells or restore mana on fallback.
* **`Battle`**: Manages turn-based combat sequencing (`play_round()`) and game loops (`simulate_battle()`) until one character wins.

```python
# --- Dice ---
d6 = Dice(sides=6)
d12 = Dice(sides=12)
print("Dice representation:", repr(d6))
print("Dice roll (d6):", d6.roll())
print("Dice roll (d12 via call):", d12())

print("-" * 40)

# --- Weapon ---
sword = Weapon("Iron Sword", d6)
axe = Weapon("Battleaxe", d12)
print("Weapon representation:", repr(sword))
print("Weapon string:", str(sword))
print("Weapon roll damage:", sword.roll_damage())
```

```python
# --- Base Character ---
dummy = Character(name="Villager", hp_max=30, damage=5)
dummy.introduce()
print("Character HP:", dummy.hp)
dummy.take_damage(10)
print("HP after 10 damage:", dummy.hp)
dummy.heal(5)
print("HP after 5 heal:", dummy.hp)

print("-" * 40)

# --- Warrior (har armor) ---
hero_warrior = Warrior(
    name="Thorin", hp_max=50, damage=10, weapon=sword, armor=3
)
print("Warrior repr:", repr(hero_warrior))
print("Warrior armor attribute:", hero_warrior.armor)

print("-" * 40)

# --- Mage (har mana) ---
hero_mage = Mage(
    name="Gandalf", hp_max=40, damage=8, weapon=None, mana=30, mana_max=30
)
print("Mage repr:", repr(hero_mage))
print("Mage mana attribute:", hero_mage.mana)
```

# 2. Combat simulations

```python
print("KAMP 1: Warrior (med våpen) VS Mage (uten våpen)")
warrior1 = Warrior(name="Thorin", hp_max=60, damage=10, weapon=axe, armor=2)
mage1 = Mage(name="Gandalf", hp_max=45, damage=6, weapon=None, mana=20)

battle1 = Battle(hero=warrior1, enemy=mage1)
winner1 = battle1.simulate_battle()
```

```python
print("\n KAMP 2: Base Character (uten våpen) VS Warrior (med våpen)")
char2 = Character(name="Goblin", hp_max=40, damage=8, weapon=None)
warrior2 = Warrior(name="Gimli", hp_max=50, damage=10, weapon=sword, armor=3)

battle2 = Battle(hero=char2, enemy=warrior2)
winner2 = battle2.simulate_battle()
```

```python
print("\n KAMP 3: Mage (med våpen) VS Warrior (uten våpen)")
mage3 = Mage(name="Saruman", hp_max=50, damage=7, weapon=sword, mana=30)
warrior3 = Warrior(name="Boromir", hp_max=55, damage=12, weapon=None, armor=4)

battle3 = Battle(hero=mage3, enemy=warrior3)
winner3 = battle3.simulate_battle()
```
