This notebook demonstrates the implementation and `pytest` test suites for **Task 1 (`Temperature`)** and **Task 2 (`Fraction`)**.

## Task 1: Temperature



```python
from temperature import Temperature

# Initialization and string representations
t_c = Temperature(25, "C")
t_f = t_c.convert_to("F")
t_k = t_c.convert_to("K")

print("str(): ", t_c, "|", t_f, "|", t_k)
print("repr():", repr(t_c), "|", repr(t_f), "|", repr(t_k))

# Equality and ordering across different units
print("25°C == 77°F:", t_c == Temperature(77, "F"))
print("0°C < 100°F: ", Temperature(0, "C") < Temperature(100, "F"))

# Validation example (below absolute zero)
try:
    Temperature(-300, "C")
except ValueError as e:
    print("Caught expected error:", e)
```

## Task 2: Fractions



```python
from fraction import Fraction

# Automatic reduction and string representations
f1 = Fraction(2, 4)
f2 = Fraction(-6, 9)
print("Reduced f1:", str(f1), "| repr:", repr(f1))
print("Reduced f2:", str(f2), "| repr:", repr(f2))

# Arithmetic operations
print("f1 + f2 =", f1 + f2)
print("f1 - f2 =", f1 - f2)
print("f1 * f2 =", f1 * f2)
print("f1 / f2 =", f1 / f2)

# Unary operations, float conversion, and comparisons
print("-f2     =", -f2)
print("abs(f2) =", abs(f2))
print("float(f1) =", float(f1))
print("1/3 < 1/2:", Fraction(1, 3) < Fraction(1, 2))
print("1/2 == 2/4:", Fraction(1, 2) == Fraction(2, 4))

# Division by zero example
try:
    _ = f1 / Fraction(0, 1)
except ZeroDivisionError as e:
    print("Caught expected error:", e)
```

```python
!pytest test_fraction.py -v
```
