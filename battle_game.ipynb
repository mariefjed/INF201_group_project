# Summary of the Game Simulator Project

## 1. Class Overview
* **`Character`**: The base class managing encapsulated health (`_hp`), damage taking, healing, attacking, and properties (`hp`, `is_alive`).
* **`Dice`**: Simulates an n-sided dice using `random.randint`, supporting direct rolling via `roll()` and instance calls (`__call__`).
* **`Weapon`**: Implements object composition by binding a name to a `Dice` object to generate dynamic weapon damage (`roll_damage()`).
* **`Warrior`**: A subclass that inherits from `Character` and overrides `take_damage()` to reduce incoming damage by `armour`.
* **`Mage`**: A subclass that manages a `mana` resource, overriding `attack()` to cast high-damage spells or restore mana on fallback.
* **`Battle`**: Manages turn-based combat sequencing (`play_round()`) and game loops (`simulate_battle()`) until one character wins.

## Combat Simulations

Below we test our classes in three distinct scenarios:
- Scenario 1: Standard characters without weapons.
- Scenario 2: Characters equipped with weapons (Dice-driven damage).
- Scenario 3: Specialised combat between a Warrior and a Mage.

```python
# Scenario 1: Basic Character Battle (No weapons)
print("=== SCENARIO 1: Basic Characters ===")
c1 = Character("Bob", hp_max=50, damage=8)
c2 = Character("Gilbert", hp_max=40, damage=10)
battle1 = Battle(c1, c2)
battle1.simulate_battle()

# Scenario 2: Weapon-equipped Battle
print("\n=== SCENARIO 2: Characters with Weapons ===")
great_sword = Weapon("Greatsword", Dice(sides=12))
dagger = Weapon("Poison Dagger", Dice(sides=6))

hero_armed = Character("Knight Arthur", hp_max=100, damage=10, weapon=great_sword)
enemy_armed = Character("Gargammel", hp_max=90, damage=10, weapon=dagger)

battle2 = Battle(hero_armed, enemy_armed)
battle2.simulate_battle()

# Scenario 3: Specialized Battle (Warrior vs Mage)
print("\n=== SCENARIO 3: Warrior vs Mage ===")
warrior_weapon = Weapon("Battle Axe", Dice(sides=8))
warrior = Warrior(name="Ironclad", hp_max=120, armour=6, weapon=warrior_weapon)
mage = Mage(name="Archmage", hp_max=80, damage=15)

battle3 = Battle(warrior, mage)
battle3.simulate_battle()
```
